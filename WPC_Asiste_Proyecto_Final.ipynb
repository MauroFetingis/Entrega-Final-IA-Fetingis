```python
import os, json, hashlib, time, sys, subprocess, re
from pathlib import Path
from datetime import datetime, timezone
from importlib.metadata import version, PackageNotFoundError
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display, Markdown

EJECUTAR_API = False
MODELO = os.environ.get('GEMINI_MODEL', 'gemini-3.8-flash')
RUTA = Path('resultados_entrega')
RUTA.mkdir(exist_ok=True)
CACHE = RUTA / 'cache_api.json'
cache = json.loads(CACHE.read_text(encoding='utf-8')) if CACHE.exists() else {}
# Recuperación de una salida real adjunta: no es una generación nueva.
# No se infieren fecha, duración o tokens de una salida sin bitácora completa.
RESPUESTAS_RECUPERADAS = json.loads('{"70c7d8040a32dacaa0d4cf1b1acf5b3da77170d8a10deec1afa834512d593253": {"etiqueta": "Zero-shot", "modelo": "gemini-3.8-flash", "prompt": "\\nCONTEXTO COMÚN DEL EXPERIMENTO ACADÉMICO:\\nEmpresa uruguaya especializada en revestimientos, decks y pérgolas WPC.\\nConsulta del cliente: Hola, estoy interesado en hacer un deck de WPC de aproximadamente 30 m² para el patio de mi casa. Me gustaría saber qué opciones tienen y cuánto podría costar.\\nDatos conocidos: deck de 30 m². Ubicación, color, fotos y plazo no informados.\\nReferencia SIMULADA de USD 45/m²; cálculo académico de USD 1350.\\nNo es una tarifa comercial ni presupuesto final. Alcance e inclusiones no confirmados.\\nSe requiere revisión humana de condiciones de instalación, estructura y acceso.\\nNo hay catálogo, stock, especificaciones técnicas ni fechas confirmadas.\\n\\n\\nRespondé como asesor comercial de una empresa de WPC a la siguiente consulta:\\n\\n\\"Hola, estoy interesado en hacer un deck de WPC de aproximadamente 30 m² para el patio de mi casa. Me gustaría saber qué opciones tienen y cuánto podría costar.\\"\\n\\nGenerá una respuesta útil para continuar la conversación con el cliente.\\n", "respuesta": "¡Hola! Muchas gracias por contactarnos. Qué excelente proyecto, el WPC es una solución ideal para patios porque brinda la calidez de la madera sin el mantenimiento constante. Con gusto te ayudamos.\\n\\nPara que tengas una primera referencia estimativa, el valor base de las tablas de deck suele rondar los **USD 45 por m²**, lo que para una superficie de 30 m² representaría un valor aproximado de **USD 1.350** (referencia inicial de material). \\n\\nTené en cuenta que este número es **meramente orientativo** y no constituye un presupuesto final. El costo definitivo dependerá de varios factores que debemos evaluar:\\n* Los accesorios de fijación (clips invisibles y tornillos de acero inoxidable).\\n* La estructura de apoyo necesaria (perfilería/alfajías).\\n* Si vas a necesitar únicamente los materiales o el servicio completo con instalación incluida.\\n* El modelo de tabla que elijas (contamos con opciones en distintas texturas y colores como teca, nogal, gris grafito, entre otros).\\n\\nPara poder asesorarte mejor y prepararte una cotización formal y precisa, ¿podrías comentarme estos detalles?\\n\\n1. **¿En qué localidad o zona de Uruguay sería la obra?**\\n2. **¿Sobre qué superficie se instalaría?** (¿ya contás con una platea/contrapiso de hormigón, o es sobre tierra/césped?).\\n3. **¿Tenés alguna foto del espacio actual?** Nos ayudaría mucho para ver accesos y desniveles.\\n4. **¿Te gustaría incluir la mano de obra o solo los materiales?**\\n\\nSi querés, podés respondernos por acá o enviarnos las fotos para avanzar con una propuesta a tu medida. ¡Quedamos a las órdenes!", "estado": "ok", "fuente": "salida_notebook_adjunto", "archivo_fuente": "WPC_Asiste_Proyecto_Final (2).ipynb", "fecha_utc": null, "segundos": null, "uso": null, "huella": "70c7d8040a32dacaa0d4cf1b1acf5b3da77170d8a10deec1afa834512d593253"}, "d5ea27236848c3f40e85742fc45ce70dcbcb8ca03a7a1662df76f427bba4e49d": {"etiqueta": "One-shot", "modelo": "gemini-3.8-flash", "prompt": "\\nCONTEXTO COMÚN DEL EXPERIMENTO ACADÉMICO:\\nEmpresa uruguaya especializada en revestimientos, decks y pérgolas WPC.\\nConsulta del cliente: Hola, estoy interesado en hacer un deck de WPC de aproximadamente 30 m² para el patio de mi casa. Me gustaría saber qué opciones tienen y cuánto podría costar.\\nDatos conocidos: deck de 30 m². Ubicación, color, fotos y plazo no informados.\\nReferencia SIMULADA de USD 45/m²; cálculo académico de USD 1350.\\nNo es una tarifa comercial ni presupuesto final. Alcance e inclusiones no confirmados.\\nSe requiere revisión humana de condiciones de instalación, estructura y acceso.\\nNo hay catálogo, stock, especificaciones técnicas ni fechas confirmadas.\\n\\n\\nActuá como asesor comercial de una empresa especializada en WPC.\\n\\n\\nEjemplo:\\n\\nConsulta:\\n\\"Quiero revestir una pared exterior con WPC, ¿qué necesitan para cotizar?\\"\\n\\nRespuesta esperada:\\n\\"¡Gracias por tu consulta! Para poder orientarte necesitamos conocer los metros cuadrados aproximados,\\nla ubicación de la obra, fotos del espacio y la terminación o color que preferís. Con esos datos podemos\\npreparar una referencia inicial. El presupuesto definitivo debe validarse luego de revisar las condiciones\\nreales de instalación.\\"\\n\\n\\nAhora respondé a esta consulta siguiendo un estilo similar:\\n\\n\\"Hola, estoy interesado en hacer un deck de WPC de aproximadamente 30 m² para el patio de mi casa. Me gustaría saber qué opciones tienen y cuánto podría costar.\\"\\n", "respuesta": "¡Hola! Gracias por contactarnos e interesarte en nuestros decks de WPC.\\n\\nPara darte una primera orientación, podemos tomar como referencia inicial un valor estimado de **USD 45 por m²**, lo que para una superficie de 30 m² rondaría los **USD 1.350**. Es fundamental tener en cuenta que este cálculo es meramente orientativo y no constituye un presupuesto final ni una tarifa comercial cerrada, ya que el alcance, los accesorios y la estructura aún deben definirse.\\n\\nPara poder presentarte las opciones adecuadas y avanzar hacia una cotización precisa, te pedimos si nos podés facilitar:\\n\\n- La **ubicación** del proyecto.\\n- Algunas **fotos o videos del patio** actual (para evaluar el tipo de suelo y el acceso).\\n- La **preferencia de color o acabado** que tenés en mente.\\n- El **plazo estimado** en el que te gustaría realizar la obra.\\n\\nCon esta información podremos orientarte mejor. Tené en cuenta que el presupuesto definitivo siempre requiere una revisión técnica de las condiciones de instalación y del terreno. \\n\\n¡Quedamos a tu disposición para coordinar los detalles!", "estado": "ok", "fuente": "salida_notebook_adjunto", "archivo_fuente": "WPC_Asiste_Proyecto_Final (2).ipynb", "fecha_utc": null, "segundos": null, "uso": null, "huella": "d5ea27236848c3f40e85742fc45ce70dcbcb8ca03a7a1662df76f427bba4e49d"}, "f9a5dd74a53406968e103474b5322c53357e2f4ffe326c67adc1040251ac813e": {"etiqueta": "Few-shot", "modelo": "gemini-3.8-flash", "prompt": "\\nCONTEXTO COMÚN DEL EXPERIMENTO ACADÉMICO:\\nEmpresa uruguaya especializada en revestimientos, decks y pérgolas WPC.\\nConsulta del cliente: Hola, estoy interesado en hacer un deck de WPC de aproximadamente 30 m² para el patio de mi casa. Me gustaría saber qué opciones tienen y cuánto podría costar.\\nDatos conocidos: deck de 30 m². Ubicación, color, fotos y plazo no informados.\\nReferencia SIMULADA de USD 45/m²; cálculo académico de USD 1350.\\nNo es una tarifa comercial ni presupuesto final. Alcance e inclusiones no confirmados.\\nSe requiere revisión humana de condiciones de instalación, estructura y acceso.\\nNo hay catálogo, stock, especificaciones técnicas ni fechas confirmadas.\\n\\nActuá como asesor comercial de WPC. Usá estos ejemplos como referencia:\\n\\nConsulta: Quiero un deck, ¿qué necesitan?\\nRespuesta: ¡Gracias por escribirnos! Contanos el metraje aproximado, ubicación, plazo y color preferido, y envianos fotos del espacio. Con eso podemos orientar la propuesta. El presupuesto final requiere revisar las condiciones de instalación.\\n\\nConsulta: Quiero revestir una fachada en altura.\\nRespuesta: ¡Gracias por tu consulta! Para avanzar necesitamos metraje, ubicación y fotos, además de altura, accesos y necesidad de andamios. Estos factores pueden modificar el costo. La solución y el presupuesto se validan con el equipo técnico.\\n\\nConsulta: ¿Cuánto cuesta una pérgola?\\nRespuesta: ¡Gracias por contactarnos! Necesitamos dimensiones, ubicación, fotos, plazo y terminación preferida. El costo depende de la estructura y de las condiciones del lugar. Podemos preparar una orientación inicial, sujeta a revisión humana; todavía no hay un presupuesto definitivo.\\n\\nAhora respondé: Hola, estoy interesado en hacer un deck de WPC de aproximadamente 30 m² para el patio de mi casa. Me gustaría saber qué opciones tienen y cuánto podría costar.", "respuesta": "¡Hola! Gracias por contactarnos. \\n\\nPara un deck de aproximadamente 30 m², a modo de referencia inicial orientativa, el valor ronda los **USD 1.350** (tomando una base estimada de USD 45/m²). Es importante aclarar que esto no constituye un presupuesto final ni una tarifa comercial confirmada, y los alcances no están definidos aún.\\n\\nPara poder presentarte opciones concretas y avanzar en una propuesta, necesitamos que nos cuentes:\\n* **Ubicación** de la obra.\\n* **Fotos del patio** para ver la superficie actual.\\n* **Preferencia de color o terminación**.\\n* **Plazo estimado** en el que te gustaría realizarlo.\\n\\nEl presupuesto definitivo dependerá de la estructura de apoyo necesaria, los accesos y las condiciones del terreno, lo cual queda siempre sujeto a la revisión técnica de nuestro equipo.", "estado": "ok", "fuente": "salida_notebook_adjunto", "archivo_fuente": "WPC_Asiste_Proyecto_Final (2).ipynb", "fecha_utc": null, "segundos": null, "uso": null, "huella": "f9a5dd74a53406968e103474b5322c53357e2f4ffe326c67adc1040251ac813e"}}')
for clave, valor in RESPUESTAS_RECUPERADAS.items():
    cache.setdefault(clave, valor)
CACHE.write_text(json.dumps(cache, ensure_ascii=False, indent=2), encoding='utf-8')
INTERVALO_SEGUNDOS = 60  # Pausa conservadora; no garantiza cuota disponible.
ultimo_intento = time.monotonic()  # También pausa antes de la primera llamada nueva.
SIMULACIONES = {'Dirigido': '¡Gracias por tu consulta! Un deck de WPC puede aportar una terminación con apariencia de madera; las prestaciones específicas deben revisarse en la ficha del producto.\n\nPara 30 m², la referencia académica de USD 45/m² da USD 1.350. Es únicamente orientativa: no es una tarifa confirmada ni define materiales, estructura o instalación incluidos.\n\nPara avanzar, contanos la ubicación, el color preferido y el plazo deseado, y envianos fotos del patio. La estructura, los accesos, la altura, los andamios y los horarios pueden modificar el costo. El presupuesto definitivo requiere revisión técnica y validación humana.', 'Iterativo': '¡Gracias por escribirnos! Para tu deck de 30 m², el cálculo académico de USD 45/m² equivale a USD 1.350. Es una referencia orientativa, sin alcance ni inclusiones confirmados; no constituye un presupuesto final.\n\n¿En qué ubicación sería la obra, qué color preferís y para cuándo la necesitás? Envianos también fotos del patio.\n\nEl equipo debe revisar la estructura, los accesos y las condiciones de instalación antes de validar el presupuesto. Las características del WPC se confirman con la ficha del producto.', 'Redes': 'Una fachada con estética de madera y terminaciones cuidadas: propuesta ilustrativa de revestimiento exterior en WPC color cedro. La resistencia y el mantenimiento dependen del producto y de su ficha técnica.\n\n#WPC #RevestimientoExterior #Terminaciones\n\n¿Querés renovar tu espacio? Escribinos para solicitar un presupuesto.'}
MODO_DEMOSTRACION = not EJECUTAR_API
registro = []
contador_consultas = 0
aciertos_cache = 0
bloquear_llamadas = False
api_key = os.environ.get('GEMINI_API_KEY', '')
if EJECUTAR_API and not api_key:
    try:
        from google.colab import userdata
        api_key = userdata.get('GEMINI_API_KEY')
    except Exception:
        api_key = ''
client = None
sdk_version = None
if EJECUTAR_API and api_key:
    try:
        import google.genai
    except ImportError:
        subprocess.check_call([sys.executable, '-m', 'pip', 'install', '-q', 'google-genai'])
    from google import genai
    sdk_version = version('google-genai')
    client = genai.Client(api_key=api_key)
print('API habilitada.' if client else 'DEMOSTRACIÓN: 3 respuestas reales recuperadas; 3 ejemplos simulados por ChatGPT, identificados como tales.')
print('Modelo:', MODELO)

def guardar_registro():
    (RUTA / 'registro_sesion.json').write_text(
        json.dumps(registro, ensure_ascii=False, indent=2), encoding='utf-8')

def consultar_modelo(prompt, etiqueta):
    global contador_consultas, aciertos_cache, bloquear_llamadas, ultimo_intento
    huella = hashlib.sha256(json.dumps([MODELO, prompt], ensure_ascii=False).encode()).hexdigest()
    if huella in cache:
        aciertos_cache += 1
        item = dict(cache[huella], origen=('salida_recuperada' if cache[huella].get('fuente') == 'salida_notebook_adjunto' else 'cache_api'), etiqueta=etiqueta)
        registro.append(item)
        guardar_registro()
        return item['respuesta']
    if MODO_DEMOSTRACION and etiqueta in SIMULACIONES:
        texto = SIMULACIONES[etiqueta]
        registro.append({'etiqueta': etiqueta, 'origen': 'simulacion_chatgpt',
                         'estado': 'demostracion', 'respuesta': texto,
                         'prompt': prompt, 'modelo': None, 'uso': None})
        guardar_registro()
        print('EJEMPLO SIMULADO POR CHATGPT. NO ES UNA RESPUESTA DE GEMINI.')
        return texto
    if client is None or bloquear_llamadas:
        print(f'{etiqueta}: PENDIENTE de ejecución real.')
        return None
    espera = max(0, INTERVALO_SEGUNDOS - (time.monotonic() - ultimo_intento))
    if espera:
        print(f'{etiqueta}: espera preventiva de {espera:.0f} segundos antes de consultar.', flush=True)
        limite = time.monotonic() + espera
        while time.monotonic() < limite:
            time.sleep(min(5, max(0, limite - time.monotonic())))
    ultimo_intento = time.monotonic()
    inicio = time.perf_counter()
    contador_consultas += 1
    item = {'etiqueta': etiqueta, 'modelo': MODELO, 'prompt': prompt,
            'fecha_utc': datetime.now(timezone.utc).isoformat(),
            'sdk_version': sdk_version, 'origen': 'api', 'huella': huella}
    try:
        interaction = client.interactions.create(model=MODELO, input=prompt)
        respuesta = interaction.output_text
        if not respuesta or not respuesta.strip():
            raise ValueError('Respuesta sin texto utilizable')
        uso = getattr(interaction, 'usage', None)
        item.update(estado='ok', respuesta=respuesta,
                    segundos=round(time.perf_counter()-inicio, 3),
                    uso=uso.model_dump(mode='json') if hasattr(uso, 'model_dump') else None)
        cache[huella] = item
        CACHE.write_text(json.dumps(cache, ensure_ascii=False, indent=2), encoding='utf-8')
    except Exception as exc:
        # Extraer sólo señales de cuota; no mostrar el mensaje crudo ni la clave.
        detalle = str(exc)
        cuerpo = getattr(exc, 'body', None)
        if cuerpo is not None:
            detalle += ' ' + str(cuerpo)
        cod = getattr(exc, 'status_code', None) or getattr(exc, 'code', None)
        if cod is None:
            cod = getattr(getattr(exc, 'response', None), 'status_code', None)
        es_limite = str(cod) == '429' or type(exc).__name__ == 'RateLimitError'
        if es_limite and cod is None:
            cod = 429
        minus = detalle.lower()
        diario = any(x in minus for x in ['perday', 'per_day', 'per day', 'daily'])
        minuto = any(x in minus for x in ['perminute', 'per_minute', 'per minute'])
        tipo_limite = 'diario' if diario else ('por_minuto' if minuto else 'no_identificado')
        metricas = sorted(set(re.findall(r'(?:Generate[A-Za-z0-9_-]*(?:PerDay|PerMinute)[A-Za-z0-9_-]*|generativelanguage[.]googleapis[.]com/[a-z_]+)', detalle)))
        retry = re.search(r'retry (?:in|after)\s+([0-9.]+)', minus)
        item.update(estado='error', tipo_error=type(exc).__name__,
                    codigo=str(cod or 'no_disponible'), tipo_limite=tipo_limite,
                    metricas_cuota=metricas,
                    espera_sugerida_segundos=float(retry.group(1)) if retry else None,
                    segundos=round(time.perf_counter()-inicio, 3))
        bloquear_llamadas = True
        print(f"{etiqueta}: error {item['codigo']} ({item['tipo_error']}). Se detienen nuevas llamadas.")
        if es_limite:
            print('Tipo de límite identificado:', tipo_limite)
            print('Métricas informadas:', metricas or 'El error no expone métricas reconocibles.')
            if retry:
                print('Espera sugerida por el proveedor (segundos):', item['espera_sugerida_segundos'])
            print('Revisar los límites activos del proyecto en Google AI Studio. No ejecutar repetidamente.')
            print('Las pausas no resuelven una cuota diaria agotada o una cuota sin disponibilidad.')
        else:
            print('Revisar permisos (401/403), modelo (404) o instalación; conservar el diagnóstico.')
        respuesta = None
    registro.append(item)
    guardar_registro()
    return respuesta

def mostrar(texto):
    if texto in SIMULACIONES.values():
        display(Markdown('**SIMULACIÓN DIDÁCTICA POR CHATGPT — NO GENERADA POR GEMINI**'))
    display(Markdown(texto)) if texto else print('PENDIENTE: no existe respuesta real para esta etapa.')
```

## Indicadores descriptivos automáticos
Se registran longitud y presencia de expresiones para orientar la revisión. No constituyen una calificación de calidad, no detectan por sí solos alucinaciones y pueden fallar ante negaciones. El límite de 150 palabras aplica al dirigido; 130 al iterativo. Las primeras tres variantes no tienen un límite explícito.

```python
import unicodedata

def normalizar(texto):
    return ''.join(x for x in unicodedata.normalize('NFD', texto.lower()) if unicodedata.category(x) != 'Mn')

def senal(texto, terminos):
    return any(x in normalizar(texto) for x in terminos)

filas = []
for tecnica, texto in respuestas.items():
    if tecnica not in ['Zero-shot', 'One-shot', 'Few-shot'] or not texto:
        continue
    limite = {'Dirigido': 150, 'Iterativo': 130}.get(tecnica)
    filas.append({'Técnica': tecnica, 'Palabras': len(texto.split()),
        'Límite explícito': limite,
        'Dentro del límite': len(texto.split()) <= limite if limite else None,
        'Señal de referencia orientativa': senal(texto, ['orientativ', 'referencia']),
        'Señal de revisión humana': senal(texto, ['valid', 'revis', 'tecnico']),
        'Menciona ubicación': senal(texto, ['ubicacion', 'donde', 'localidad', 'zona']),
        'Menciona fotos': senal(texto, ['foto', 'imagen']),
        'Menciona color': senal(texto, ['color', 'terminacion']),
        'Menciona plazo': senal(texto, ['plazo', 'fecha'])})
validacion_api = pd.DataFrame(filas)
if not validacion_api.empty:
    display(validacion_api)
    validacion_api.to_csv(RUTA / 'indicadores.csv', index=False)
    fig, ax = plt.subplots(figsize=(9, 4))
    ax.bar(validacion_api['Técnica'], validacion_api['Palabras'], color='#126c73')
    ax.set(ylabel='Palabras', title='Extensión de las respuestas reales (no mide calidad)')
    fig.tight_layout()
    fig.savefig(RUTA / 'extension_respuestas.png', dpi=160)
    plt.show()
else:
    print('Indicadores y gráfico pendientes: todavía no hay respuestas reales.')
```
